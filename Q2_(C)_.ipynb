## **Solution to Q.2.(C)**

```python
import numpy as np

# =========================================================
# Q2(c)
# Evaluate Pi_N(c), grad Pi_N(c), and Hessian Pi_N(c)
# using analytical expressions
# =========================================================

def PiN(c):

    c1, c2 = c

    return (
        c1**2 / 2
        + c1*c2
        + 2*c2**2 / 3
        + c1**4 / 20
        + c1**3*c2 / 5
        + 3*c1**2*c2**2 / 14
        + c1*c2**3 / 8
        + c2**4 / 36
        - c1 / 2
        - c2 / 3
    )


def grad_PiN(c):

    c1, c2 = c

    dPi_dc1 = (
        c1
        + c2
        + c1**3 / 5
        + c1**2*c2 / 2
        + 3*c1*c2**2 / 7
        + c2**3 / 8
        - 1/2
    )

    dPi_dc2 = (
        c1
        + 4*c2 / 3
        + c1**3 / 6
        + 3*c1**2*c2 / 7
        + 3*c1*c2**2 / 8
        + c2**3 / 9
        - 1/3
    )

    return np.array([
        dPi_dc1,
        dPi_dc2
    ])


def hess_PiN(c):

    c1, c2 = c

    H11 = (
        1
        + 3*c1**2 / 5
        + c1*c2
        + 3*c2**2 / 7
    )

    H12 = (
        1
        + c1**2 / 2
        + 6*c1*c2 / 7
        + 3*c2**2 / 8
    )

    H21 = H12

    H22 = (
        4/3
        + 3*c1**2 / 7
        + 3*c1*c2 / 4
        + c2**2 / 3
    )

    return np.array([
        [H11, H12],
        [H21, H22]
    ])


# =========================================================
# Evaluate at c = [1, -0.5]
# =========================================================

c = np.array([1.0, -0.5])

print("c =", c)

print("\nPi_N(c) =")
print(PiN(c))

print("\nGradient grad(Pi_N) =")
print(grad_PiN(c))

print("\nHessian grad^2(Pi_N) =")
print(hess_PiN(c))
```
